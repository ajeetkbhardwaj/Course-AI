### Introduction to Audio
* **Sound waves are continuous signals** — they contain infinite values over time.
* **Digital devices require discrete (finite) data**, so continuous sound must be converted into a digital form.
* The **conversion process** from analog (continuous) to digital involves **sampling** (taking measurements at discrete time intervals).
* This conversion is done by an **Analog-to-Digital Converter (ADC)**.
* **Audio datasets** store these digital representations as files (not continuous waves).
* Common **audio file formats** include:

  * `.wav` — uncompressed or lossless format.
  * `.flac` — compressed but lossless.
  * `.mp3` — compressed and lossy.
* The **main difference between formats** lies in how they **compress** and **store** the digital signal.
* The **signal capture process**:

  1. Sound wave → captured by **microphone** → electrical signal.
  2. Electrical signal → **ADC** → digital representation (samples).

What is sampling and sampling rate ?
Sampling - a process of measuring values of continuous signal at fixed time steps. The measured discrete signal value are stored into the specific format like wave format with it's sampling rate.

![alt text](https://upload.wikimedia.org/wikipedia/commons/thumb/c/c3/Signal_Sampling.svg/500px-Signal_Sampling.svg.png)

Sampling rate(frequency) - The number of samples taken in one second  and measured in Hz. 

Ex-1 : CD-quality audio has a sampling rate of 44,100 Hz. High resoultion audio has sampling rate of 192,000 Hz or 192 kHz. The common sampling rate used to speech model training is 16,000 Hz or 16 kHz.

**1. Sampling Rate Basics**

* **Sampling rate** = number of samples taken per second (in Hz).
* Determines the **highest frequency** that can be captured — called the **Nyquist limit** (½ of the sampling rate).

**2. Practical Examples**

* **Human speech frequencies**: below **8 kHz** → **16 kHz sampling rate** is sufficient.
* **Too high** a rate = no extra information + higher computational cost.
* **Too low** a rate = **information loss** → muffled or poor-quality audio.

  * Example: Speech at **8 kHz** sounds muffled because high frequencies are lost.

**3. Consistency in Sampling Rate**

* All audio samples in a dataset should have the **same sampling rate**.
* When **fine-tuning** a pre-trained model, match your data’s sampling rate to the **model’s pre-training rate**.

**4. Impact on Data Representation**

* Sampling rate affects **temporal resolution** and **sequence length**:

  * 5 seconds @ **16,000 Hz** → **80,000 samples**.
  * 5 seconds @ **8,000 Hz** → **40,000 samples**.
* Different sampling rates = different sequence lengths → models (especially **Transformers**) may struggle to generalize.

**5. Resampling**

* **Resampling** = adjusting all audio to the **same sampling rate**.
* Essential preprocessing step for **consistent model training and evaluation**.

### 🎵 **Amplitude and Bit Depth**

* **Sampling rate** → how *often* samples are taken.

* **Amplitude** → how *strong* the sound wave is at a given moment; represents **sound pressure level**.

* **Amplitude is measured in decibels (dB)** and perceived as **loudness**.

  * Normal speaking voice ≈ **< 60 dB**.
  * Rock concert ≈ **125 dB**.

* Each **audio sample** stores the **amplitude** of the wave at a specific point in time.

* **Bit depth** → determines how precisely the amplitude is represented (resolution of the sample).

* **Higher bit depth** = more accurate representation + less **quantization noise**.

* **Common bit depths:**

  * **16-bit** → 65,536 possible amplitude levels.
  * **24-bit** → 16,777,216 possible amplitude levels.

* **Quantization noise** = rounding error when converting continuous values to discrete ones.

* **Higher bit depth** → smaller quantization noise → better audio quality.

* **16-bit audio** already has inaudible quantization noise → sufficient for most uses.

* **32-bit audio** → uses **floating-point samples** (range: **[-1.0, 1.0]**).

  * Has **24-bit precision**, same as 24-bit integer audio.
  * Useful because **machine learning models** operate on **floating-point data**, so audio must be **converted** to this format.

* **Amplitude in digital audio** is still expressed in **decibels (dB)**.

* **Human hearing is logarithmic**, so **decibel scale** makes loudness easier to interpret.

* **Decibel reference points:**

  * Real-world audio: **0 dB** = quietest audible sound.
  * Digital audio: **0 dB** = *maximum possible amplitude* (everything else is negative).

* **Quick rule:**

  * Every **-6 dB** → amplitude halves.
  * Below **-60 dB** → generally inaudible.
